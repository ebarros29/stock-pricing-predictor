# NVDA Stock — Exploratory Data Analysis
Symbol: NVDA | Period: 2018-01-01 to 2024-07-20

```python
import yfinance as yf
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use('seaborn-v0_8-darkgrid')
sns.set_palette('husl')
```

```python
symbol = 'NVDA'
df = yf.download(symbol, start='2018-01-01', end='2024-07-20', auto_adjust=True)
if isinstance(df.columns, pd.MultiIndex):
    df.columns = df.columns.droplevel(1)
df.columns = [c.strip().replace(' ', '_') for c in df.columns]
print(f'Shape: {df.shape}')
df.head()
```

## Closing Price Over Time

```python
fig, ax = plt.subplots(figsize=(14, 5))
ax.plot(df.index, df['Close'], linewidth=0.8)
ax.set_title(f'{symbol} — Adjusted Close Price', fontsize=14)
ax.set_ylabel('Price (USD)')
plt.tight_layout()
plt.show()
```

## Moving Averages (20 & 50 days)

```python
df['MA20'] = df['Close'].rolling(20).mean()
df['MA50'] = df['Close'].rolling(50).mean()

fig, ax = plt.subplots(figsize=(14, 5))
ax.plot(df.index, df['Close'], linewidth=0.8, alpha=0.5, label='Close')
ax.plot(df.index, df['MA20'], linewidth=1, label='MA 20')
ax.plot(df.index, df['MA50'], linewidth=1, label='MA 50')
ax.set_title(f'{symbol} — Price with Moving Averages', fontsize=14)
ax.legend()
plt.tight_layout()
plt.show()
```

## Daily Returns Distribution

```python
df['Daily_Return'] = df['Close'].pct_change() * 100

fig, ax = plt.subplots(figsize=(12, 5))
ax.hist(df['Daily_Return'].dropna(), bins=80, alpha=0.75, edgecolor='white')
ax.set_title(f'{symbol} — Daily Returns Distribution', fontsize=14)
ax.set_xlabel('Return (%)')
ax.set_ylabel('Frequency')
plt.tight_layout()
plt.show()
```

```python
print(f'Mean daily return: {df["Daily_Return"].mean():.3f}%')
print(f'Std daily return:  {df["Daily_Return"].std():.3f}%')
print(f'Skewness:          {df["Daily_Return"].skew():.3f}')
```

## Volume Over Time

```python
fig, ax = plt.subplots(figsize=(14, 4))
ax.fill_between(df.index, df['Volume'], alpha=0.3)
ax.set_title(f'{symbol} — Trading Volume', fontsize=14)
ax.set_ylabel('Volume')
plt.tight_layout()
plt.show()
```

## Feature Correlation Heatmap

```python
corr = df[['Open', 'High', 'Low', 'Close', 'Volume']].corr()
fig, ax = plt.subplots(figsize=(8, 6))
sns.heatmap(corr, annot=True, fmt='.3f', cmap='coolwarm', center=0, ax=ax)
ax.set_title(f'{symbol} — Feature Correlations', fontsize=14)
plt.tight_layout()
plt.show()
```
